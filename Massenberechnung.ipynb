```python
import numpy as np

def MassOfSphere(density, diameter):
    volume = 4/3 * np.pi * (diameter / 2000)**3
    mass = density * 1000 * volume
    return mass

density_SS304 = 7.930 # Dichte Edelstahl 304 in g/cm^3
diameters_sphere = [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6, 7, 8] # Durchmesser in mm

# mass = MassOfSphere(density_SS304, diameters_sphere)

for i in diameters_sphere:
    mass = MassOfSphere(density_SS304, i)
    print(f"{i}mm Sphere has a Weight of {round(mass*1000000)/1000}g")
```
