```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
beta = 500
a = 0.013
m = 0.00127

# k-Werte (ohne exakt 0, um Division zu vermeiden)
k = np.linspace(-np.pi/a, np.pi/a, 1000)
k[k == 0] = 1e-10  # numerischer Trick

# Grundfrequenz
omega0 = 2 * np.sqrt(beta / m)

# Dispersionsrelation
omega = omega0 * np.abs(np.sin(k * a / 2))

# 1. Schallgeschwindigkeit (konstant)
c = omega0 * a / 2

# 2. Phasengeschwindigkeit
v_ph = omega / k

# 3. Gruppengeschwindigkeit
sin_term = np.sin(k * a / 2)
cos_term = np.cos(k * a / 2)

# sign(sin) für korrekte Richtung
sign = np.sign(sin_term)

v_gr = (a * omega0 / 2) * cos_term * sign

# Plot
plt.figure(figsize=(10, 6))

plt.plot(k, v_ph, label=r"$v_{\mathrm{Ph}}(k)$", linewidth=2, color = "red")
plt.plot(k, v_gr, label=r"$v_{\mathrm{Gr}}(k)$", linewidth=2, color = "blue")
plt.axhline(c, linestyle='--', label=r"$c$", linewidth=2, color="green")

plt.xlabel("Wellenzahl $k$")
plt.ylabel("Geschwindigkeit [m/s]")
plt.title("Phasen-, Gruppen- und Grenzgeschwindigkeit")
plt.legend()
plt.grid(True)

plt.xticks(
    ticks=[-np.pi/a, 0, np.pi/a],
    labels=[r"$-\dfrac{\pi}{a}$", r"$0$", r"$\dfrac{\pi}{a}$"],
    fontsize=14
)

plt.tight_layout()
plt.show()
```
