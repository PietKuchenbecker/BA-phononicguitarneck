```python
import numpy as np
import matplotlib.pyplot as plt

# Gegebene Parameter

beta = 500  # N/m, typische Federkonstante für Silikon
a = 0.013  # Gitterabstand in m

k = np.linspace(-np.pi/(a), np.pi/(a), 1000) # Wellenzahlen

m = 0.00127   # aim for 200 Hz

# Dispersionsrelation berechnen

omega0 = 2 * np.sqrt(beta / m)

omega = omega0 * np.abs(np.sin(k * a / 2))

# In Frequenzbereich umrechnen

f_omega = omega / (2*np.pi)

# Plot

plt.figure(figsize=(10, 6))
plt.plot(k, f_omega, 'b-', label="Akustischer Zweig")
plt.title("Dispersionsrelation für 1D monoatomischen Kristall")
plt.xlabel("Wellenzahl $k$")
plt.ylabel("Frequenz [Hz]")
plt.legend()
plt.grid(True)

plt.xticks(
ticks=[-np.pi/(a), 0, np.pi/(a)],
labels=[r"$-\frac{\pi}{a}$", r"$0$", r"$\frac{\pi}{a}$"],
fontsize=14
)

plt.tight_layout()
plt.show()
```
